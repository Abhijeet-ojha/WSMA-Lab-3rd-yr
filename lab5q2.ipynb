```
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction.text import CountVectorizer
```

```
articles = [
    "Artificial intelligence is changing technology. AI helps businesses improve technology.",
    "Football players practise daily. Football improves fitness and teamwork.",
    "Students use technology for online education. Education improves knowledge.",
    "Doctors use technology to improve healthcare. Healthcare needs skilled doctors."
]
```

```
# 1. Lexical diversity
for i, text in enumerate(articles):
    words = text.lower().split()
    diversity = len(set(words)) / len(words)
    print("Article", i+1, "Lexical Diversity:", round(diversity, 2))
```

```
for n in [1, 2, 3]:
    cv = CountVectorizer(ngram_range=(n, n), stop_words="english")
    data = cv.fit_transform(articles)
    print("\nN-grams:", n)
    print(cv.get_feature_names_out())
```

```
# 3. TF-IDF model
tfidf = TfidfVectorizer(stop_words="english")
matrix = tfidf.fit_transform(articles)
terms = tfidf.get_feature_names_out()
```

```
# 4. Top 5 keywords from each article
for i in range(len(articles)):
    scores = matrix[i].toarray().flatten()
    top = scores.argsort()[-5:][::-1]
    print("\nArticle", i+1, "Top Keywords:")
    for j in top:
        if scores[j] > 0:
            print(terms[j], round(scores[j], 3))
```

```
# 5. Compare articles
print("\nARTICLE COMPARISON")
for i, text in enumerate(articles):
    words = text.lower().split()
    print("Article", i+1,
          "Words:", len(words),
          "Unique:", len(set(words)))
```
