```python
import nltk
import re
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
text="""
Here is a selection of random text paragraphs mixed with emojis, tailored for different moods and ready to copy and paste.🚀 Creative ChaosThe hyper-loop train just bypassed Mars 🪐 because someone left a half-eaten burrito 🌯 on the control panel. Now the AI pilot 🤖 is singing opera hits from the 1920s while raining neon jellybeans 🍬 all over the virtual lounge. Grab your laser-umbrella ☂️ and dance through the pixel storm before the gravity grid resets 🌀!☕ Cozy SundaySoft sunlight filters through the window 🪟 as a warm mug of freshly brewed coffee ☕ steams on the table. A sleepy cat 🐱 curls up tightly on a fluffy blanket, purring like a tiny, gentle engine. Outside, autumn leaves 🍂 drift slowly to the pavement, creating a quiet afternoon perfect for reading 📚.🕹️ Gamer EnergyThe final boss emerged from the digital fog 🌫️ with three health bars and a glowing sword ⚔️. My headset was buzzing with static 🎧 while my teammates screamed completely conflicting strategies into the chat. With only 2% health left 🩸 and an empty inventory, I closed my eyes, mashed every single button 🎮, and somehow unlocked a secret achievement 🏆.
"""
text = text.lower()
text = re.sub(r'http\S+', '', text)
text = re.sub(r'[^\x00-\x7f]+','',text)
```

```python
tokens = word_tokenize(text)
tokens =[
    word for word in tokens if word not in string.punctuation
]
stop_words = set(stopwords.words('english'))
clean_tokens=[
    word for word in tokens if word not in stop_words
]

print("Original tokens\n")
print(tokens)

print("\nCleaned tokens\n")
print(clean_tokens)
```

## Student feedback anlysis


```python
text_samples = [
    "The workshop was very informative 😊. Please upload the slides at https://college.edu. Thank you!!",
    "Great session on AI, loved the hands-on practice! 👍 Check this out: http://ai-workshop.com 🔥",
    "Omg it was wayyy too long... 😴 but the instructor was nice i guess. email me at student@uni.edu.",
    "Highly recommended!!! 🌟 10/10 stars. register for next week here: https://linktr.ee",
    "I could not hear the audio properly at the back 😢. can you please fix the mic next time? thx.",
    "the coding exercises were a bit tough 🤯 but useful. Slides link? www.collegedata.org/slides/python"
]
cleaned_text_samples = []
for text in text_samples:
    # Lowercase
    text_lower = text.lower()
    # Remove URLs (handles http, https, and www)
    text_no_url = re.sub(r'https?://\S+|www\.\S+', '', text_lower)
    # Remove non-ASCII (emojis)
    text_no_emoji = re.sub(r'[^\x00-\x7f]+', '', text_no_url)
    
    cleaned_text_samples.append(text_no_emoji)
```

```python
og_tokens=[]
cleaned_tokens =[]
for text in cleaned_text_samples:
    tokens = word_tokenize(text)
    tokens =[
        word for word in tokens if word not in string.punctuation
    ]
    stop_words = set(stopwords.words('english'))
    clean_tokens=[
        word for word in tokens if word not in stop_words
    ]

    og_tokens.append(tokens)
    cleaned_tokens.append(clean_tokens)
print("Original tokens/n")
for token in og_tokens:
    print(token)
print("Cleaned tokens/n")
for token in cleaned_tokens:
    print(token)
```

```python
#implementing the same using Scapy

import spacy
import re
nlp = spacy.load("en_core_web_sm")
for text in cleaned_text_samples:
    doc = nlp(text)
    clean_tokens = []
    for token in doc:
        if not token.is_stop and not token.is_punct and token.text.strip() != "":
            clean_tokens.append(token.text)
    print(clean_tokens)
```
