```
import re
import pandas as pd
from datetime import datetime
```

```
log_data = """
127.0.0.1 - - [10/Oct/2023:13:55:36 +0000] "GET /index.html HTTP/1.1" 200 1024
192.168.0.5 - - [10/Oct/2023:13:58:12 +0000] "GET /about.html HTTP/1.1" 200 204
127.0.0.1 - - [10/Oct/2023:14:05:45 +0000] "POST /login HTTP/1.1" 302 512 "http
"""
log_pattern = re.compile(
r'(\S+) - - \[(.*?)\] "(\S+) (\S+) \S+" (\d{3}) \d+ "(.*?)" ".*?"'
)
records = []
for line in log_data.strip().split("\n"):
  match = log_pattern.match(line)
  if match:
    ip, timestamp, method, url, status, referrer = match.groups()
  # Convert timestamp string to datetime object
    timestamp = datetime.strptime(timestamp.split()[0], "%d/%b/%Y:%H:%M:%S")
    records.append([ip, timestamp, method, url, status, referrer])
```

```
# Step 5: Create a DataFrame for analysis (Added "Referrer" to columns)
df = pd.DataFrame(records, columns=["IP", "Timestamp", "Method", "URL", "Status", "Referrer"])
print(df)

# Step 6: Find Top IP Addresses
print(df["IP"].value_counts())

# Step 7: Find Top Referrers
print(df["Referrer"].value_counts())

# Step 8: Approximate Session Duration per IP (Fixed timestamp conversion and typo)
df["Timestamp"] = pd.to_datetime(df["Timestamp"])  # Ensure timestamps are datetime objects
session_duration = df.groupby("IP")["Timestamp"].agg(lambda x: (max(x) - min(x)).total_seconds())
print(session_duration)
```
